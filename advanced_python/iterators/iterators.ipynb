## 📝 4. Detailed Explanations
An iterator is an object that implements the iteration protocol, which consists of the methods `__iter__()` (which returns the iterator itself) and `__next__()` (which returns the next element). When there are no more elements, `__next__()` raises a `StopIteration` exception.

---

## ⚠️ 8. Common Mistakes & Pitfalls
Be on the lookout for these frequent errors:
- **Pitfall:** Forgetting to raise StopIteration in custom iterators, which creates infinite loops during iteration.\n- **Pitfall:** Assuming an iterator can be reset or reused; once an iterator is exhausted, it will raise StopIteration on subsequent calls, and a new iterator must be created.

## 💼 9. Interview Questions
Expected conceptual and coding questions in technical interviews:
- Q: What is the difference between an iterable and an iterator?
A: An iterable is any object that can return an iterator via its __iter__ method (e.g. lists, tuples). An iterator is the object that actually performs the iteration by returning elements one-by-one via its __next__ method.\n- Q: How does a for loop know when to stop iterating?
A: It catches the StopIteration exception raised by the iterator's __next__ method and terminates the loop cleanly.

## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
class CountDown:
    def __init__(self, start):
        self.current = start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        self.current -= 1
        return self.current + 1
print('Custom countdown iterator:', list(CountDown(3)))
```
