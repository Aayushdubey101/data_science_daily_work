## 📝 4. Detailed Explanations
A generator is a special type of iterator defined using a standard function containing the `yield` statement. Unlike regular functions that run to completion and return a value, generator functions pause execution and yield values one-by-one, preserving state across calls.

---

## ⚠️ 8. Common Mistakes & Pitfalls
Be on the lookout for these frequent errors:
- **Pitfall:** Calling the generator function multiple times inside loop headers (e.g. `next(lazy_range(3))`), which recreates the generator instance every time, resulting in an infinite loop yielding the first item.\n- **Pitfall:** Attempting to slice a generator directly like a list (generators are non-subscriptable; use `itertools.islice` instead).

---

## 💼 9. Interview Questions
Expected conceptual and coding questions in technical interviews:
- Q: What happens to a generator function's state when it hits a yield statement?
A: The function's execution state, local variables, and instruction pointer are paused and preserved, yielding control and the value back to the caller.\n- Q: What is the main advantage of generators over lists for large datasets?
A: Memory efficiency. Generators evaluate items lazily on-demand, requiring O(1) memory, whereas lists allocate all elements in memory simultaneously, requiring O(N) memory.

## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
def lazy_range(limit):
    count = 0
    while count < limit:
        yield count
        count += 1
gen = lazy_range(3)
print('First item:', next(gen))
print('Remaining items:', list(gen))
```
