## 💡 2. Why This Topic Matters

## 🧠 3. Core Concepts
Here are the essential terms and key pillars to master:
- **Extending the base Exception class.**\n- **Custom error attributes and logging messages.**\n- **Raising custom errors inside validation loops.**

---

## 📝 4. Detailed Explanations
Custom exceptions are defined by creating a class that inherits from the built-in `Exception` class. This allows you to write domain-specific error types that can be caught and handled explicitly.

---

## ⚠️ 8. Common Mistakes & Pitfalls
Be on the lookout for these frequent errors:
- **Pitfall:** Inheriting from `BaseException` instead of the standard `Exception` class (BaseException catches system-level events that should not be intercepted by user code).\n- **Pitfall:** Creating too many custom exceptions for trivial bugs that could be represented by standard built-in exceptions like `ValueError`.

---

## 💼 9. Interview Questions
Expected conceptual and coding questions in technical interviews:
- Q: Why would you define a custom exception instead of using standard exceptions?
A: To provide domain-specific context, clean error categorization, and allow business logic errors to be handled separately from standard system bugs.\n- Q: How do you define a custom exception class in Python?
A: Create a subclass inheriting from the built-in 'Exception' class and initialize it with your custom attributes and message.

## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
class OutOfStockError(Exception):
    def __init__(self, item_name, msg='Requested item is currently out of stock.'):
        self.item_name = item_name
        super().__init__(f'{msg} Item: {item_name}')

try:
    raise OutOfStockError('RTX-4090')
except OutOfStockError as e:
    print('Caught Custom Exception:', e.item_name)
```
