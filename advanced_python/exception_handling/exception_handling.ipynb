---

## 📝 4. Detailed Explanations
Exceptions are runtime errors that halt script execution. The `try` block wraps code that might raise an error. `except` intercepts specific exceptions. `else` runs if no exceptions occur, and `finally` runs unconditionally, making it ideal for cleaning up resources.

---

## ⚠️ 8. Common Mistakes & Pitfalls
Be on the lookout for these frequent errors:
- **Pitfall:** Using bare `except:` statements, which catch everything including system exits and keyboard interrupts, making debugging difficult.\n- **Pitfall:** Not specifying narrow, explicit exception types (e.g. catching `Exception` instead of `FileNotFoundError`).

---

## 💼 9. Interview Questions
Expected conceptual and coding questions in technical interviews:
- Q: What is the purpose of the 'else' block in exception handling?
A: It executes only if the code inside the try block runs successfully without raising any exceptions.\n- Q: Why is it bad practice to write a bare except: statement?
A: It masks unexpected errors and prevents keyboard interrupts (`Ctrl+C`) or system exits from terminating the script, making the application difficult to debug.

## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
try:
    res = 10 / 2
except ZeroDivisionError:
    print('Zero denominator!')
else:
    print('Calculated result:', res)
finally:
    print('Teardown routine completed.')
```
