## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
# Simulating a retail billing checkout terminal
class Product:
    def __init__(self, name, price):
        self.name = name
        self.__price = price # encapsulated
    @property
    def price(self):
        return self.__price
class Bill:
    def __init__(self):
        self.items = []
    def add_product(self, product, qty):
        self.items.append((product, qty))
    def calculate_total(self):
        return sum(prod.price * qty for prod, qty in self.items)
# Demo
p1 = Product('Notebook', 4.5)
b = Bill()
b.add_product(p1, 2)
print('Total retail checkout cost:', b.calculate_total())
```
