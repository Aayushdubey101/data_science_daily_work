## 📝 4. Detailed Explanations
A class is a user-defined blueprint containing attributes (state) and methods (behavior). An object is an instance of a class with concrete data populated into its fields.

## ⚠️ 8. Common Mistakes & Pitfalls
Be on the lookout for these frequent errors:
- **Pitfall:** Modifying class attributes through instances, which can create accidental instance attributes with the same name.\n- **Pitfall:** Confusing instance methods with regular functions and omitting the mandatory `self` argument.

## 💼 9. Interview Questions
Expected conceptual and coding questions in technical interviews:
- Q: What is the difference between a class attribute and an instance attribute?
A: A class attribute is shared by all instances of a class. An instance attribute is unique to each specific object instance.\n- Q: What does instantiation mean?
A: The process of creating a concrete object instance from a class blueprint.
