## 💡 2. Why This Topic Matters

## 🧠 3. Core Concepts
Here are the essential terms and key pillars to master:
- **String representations: __str__ vs __repr__.**\n- **Operator overloading (__add__, __len__).**\n- **Context management magic methods (__enter__, __exit__).**

---

## 📝 4. Detailed Explanations
Magic methods (dunder methods, starting/ending with double underscores) allow custom objects to hook into Python's built-in syntax. For example, overriding `__str__` customizes user-facing print output, while `__repr__` customizes developer-facing debugging output.

---

## ⚠️ 8. Common Mistakes & Pitfalls
Be on the lookout for these frequent errors:
- **Pitfall:** Overriding magic methods without returning the correct object types, violating standard Python behavior.\n- **Pitfall:** Confusing `__str__` (user-friendly display) with `__repr__` (unambiguous developer representation).

## 💼 9. Interview Questions
Expected conceptual and coding questions in technical interviews:
- Q: What is the difference between __str__ and __repr__?
A: __str__ returns a user-friendly string representation of the object. __repr__ returns an unambiguous string representation meant for debugging, which ideally could be used to recreate the object.\n- Q: How do you overload the '+' operator for a custom class?
A: Override the __add__(self, other) magic method.

## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __repr__(self):
        return f'Vector({self.x}, {self.y})'
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
v1 = Vector(2, 3)
v2 = Vector(4, 5)
print('Added Vectors:', v1 + v2)
```
