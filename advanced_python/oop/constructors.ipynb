---

## 📝 4. Detailed Explanations
`__new__` is the actual constructor method responsible for allocating memory and returning a new instance of the class. `__init__` is the initializer method that configures the newly created instance.

---

## ⚠️ 8. Common Mistakes & Pitfalls
Be on the lookout for these frequent errors:
- **Pitfall:** Forgetting to return the instance from `__new__`, which prevents `__init__` from executing and leaves the variable bound to `None`.\n- **Pitfall:** Confusing `__init__` as the actual memory allocator (it only initializes existing instances).

## 💼 9. Interview Questions
Expected conceptual and coding questions in technical interviews:
- Q: What is the primary difference between __new__ and __init__?
A: __new__ is a static method that allocates memory and returns a new object instance. __init__ is an instance method that initializes the attributes of that instance.\n- Q: How do you implement a Singleton class in Python?
A: Override the __new__ method to track and return a single shared instance variable.

## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
class Singleton:
    _instance = None
    def __new__(cls, *args, **kwargs):
        if not cls._instance:
            cls._instance = super().__new__(cls)
        return cls._instance
s1 = Singleton()
s2 = Singleton()
print('Are s1 and s2 the exact same instance?', s1 is s2)
```
