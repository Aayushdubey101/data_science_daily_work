## 📝 4. Detailed Explanations
A decorator is a function that takes another function as an argument, extends its behavior, and returns a modified function wrapper. Using `@functools.wraps` inside the decorator preserves the original function's name and docstring.

---

## ⚠️ 8. Common Mistakes & Pitfalls
Be on the lookout for these frequent errors:
- **Pitfall:** Forgetting to use `@functools.wraps` inside decorators, which overwrites the decorated function's metadata (`__name__` and `__doc__`) with the wrapper's details.\n- **Pitfall:** Forgetting to return the result value from the inner wrapper function, which causes the decorated function to return `None` silently.

## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
import functools
import time
def execution_timer(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        print(f'{func.__name__} took {time.time() - start:.4f}s')
        return result
    return wrapper
@execution_timer
def process_data(): time.sleep(0.1)
process_data()
```
