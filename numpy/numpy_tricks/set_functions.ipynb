## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
import numpy as np
a = np.array([1, 2, 2, 3])
b = np.array([3, 4, 5])
print('Unique items in a:', np.unique(a))
print('Intersection:', np.intersect1d(a, b))
print('Difference (a - b):', np.setdiff1d(a, b))
```
