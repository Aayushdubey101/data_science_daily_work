## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
import pandas as pd
idx = pd.MultiIndex.from_tuples([('A', 1), ('B', 1)])
s = pd.Series([10, 20], index=idx)
unstacked = s.unstack()
print('Unstacked:\n', unstacked)
print('Stacked back:\n', unstacked.stack())
```
