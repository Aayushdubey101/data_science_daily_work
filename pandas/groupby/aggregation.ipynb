## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
import pandas as pd
df = pd.DataFrame({'Dept': ['HR', 'IT', 'IT'], 'Salary': [50, 80, 90], 'Bonus': [5, 10, 15]})
agg_df = df.groupby('Dept').agg({'Salary': 'mean', 'Bonus': ['sum', 'max']})
print('Aggregations:\n', agg_df)
```
