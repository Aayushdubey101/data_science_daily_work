## 💻 5. Multiple Code Examples & Real-world Scenarios
Below are clean, commented execution examples demonstrating the concepts in action:


```python
import pandas as pd
df = pd.DataFrame({'Year': [2020, 2020, 2021], 'Sector': ['Tech', 'Bio', 'Tech'], 'Sales': [100, 150, 120]})
pivot = df.pivot_table(values='Sales', index='Year', columns='Sector', aggfunc='sum', fill_value=0)
print('Pivot Table:\n', pivot)
```
